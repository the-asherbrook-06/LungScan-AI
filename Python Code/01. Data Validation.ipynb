# Importing Packages

````python
import pandas as pd
import os
````

# Defining Dataset

````python
DATASET_PATH = "Dataset"
IMAGE_PATH = os.path.join(DATASET_PATH, "Images")
````

# Importing Dataset

````python
labels = pd.read_csv(os.path.join(DATASET_PATH, "01. labels.csv"))
````

````python
with open(os.path.join(DATASET_PATH, "01. train_val.txt")) as f:
    train_val = set(f.read().splitlines())
````

````python
with open(os.path.join(DATASET_PATH, "01. test.txt")) as f:
    test = set(f.read().splitlines())
````

````python
downloaded_images = set(os.listdir(IMAGE_PATH))
````

# List Splits

Since we have only downloaded ~15k images, we must work only with those. The `train_val.txt` and `test.txt` contains all entries from the original dataset, so we need to remove the entries which we haven't downloaded.

````python
train_val_subset = downloaded_images.intersection(train_val)
test_subset = downloaded_images.intersection(test)
````

````python
print("Train/Val subset Length:\t", len(train_val_subset))
print("Test subset Length:\t\t", len(test_subset))
print("Train/Val subset Percent:\t", (len(train_val_subset) / len(downloaded_images)*100))
print("Test subset Percent:\t\t", (len(test_subset) / len(downloaded_images)*100))
````

Since the Dataset has **~84%** for Training + Validation and **~16%** for Test Dataset, we can proceed with the project development without any issues

# Patient Grouping

````python
subset_labels = labels[labels["Image Index"].isin(downloaded_images)]
````

````python
train_val_labels = subset_labels[subset_labels["Image Index"].isin(train_val_subset)]
test_labels = subset_labels[subset_labels["Image Index"].isin(test_subset)]
````

````python
train_val_patients = set(train_val_labels["Patient ID"])
test_patients = set(test_labels["Patient ID"])
````

````python
print("Unique Train/Val patients:\t", len(train_val_patients))
print("Unique Test patients:\t\t", len(test_patients))
````

Since we have **~150** Unique Test patients and **~3800** Train + Validation Patients, we can proceed further without any issues

# Patient Overlap Check

````python
patient_overlap = train_val_patients.intersection(test_patients)
````

````python
print("Patients appearing in BOTH sets:", len(patient_overlap))
````

Since we don't have any patients overlapping in both Train + Validation and Test subsets, we are good to proceed further

# Images per Patient

````python
patient_image_counts = subset_labels.groupby("Patient ID")["Image Index"].count()
````

````python
print("Average images per patient:\t", patient_image_counts.mean())
````

Since the Dataset has a average of **~4** Images per patient, the model can pick up patterns better from each patients...

# Saving Subset List

Since we have the reduced, cleaned and focused Train/Val and Test List, we can save it into seperate files for easy access

````python
with open("Dataset/02. train_val_subset.txt", "w") as f:
    for img in train_val_subset:
        f.write(img + "\n")
````

````python

with open("Dataset/02. test_subset.txt", "w") as f:
    for img in test_subset:
        f.write(img + "\n")
````
