# Dataset Structure

The Dataset for this project is collected from [NIH Chest X-rays](https://www.kaggle.com/datasets/nih-chest-xrays/data) dataset from Kaggle. The Original Dataset contains over 112000 Chest X-ray images from more than 30,000 unique patients, amounting to approx. 45Gb in size.

So for this project, we have sampled only the first 15000 images to train the model and do the predictions, out of which 25 entries were randomly sampled and verified with an local doctor and physician, proving the dataset's credibility.

The collected Dataset has the following file structure:
```bash
Dataset/
- labels.csv
- train_val.txt
- test.text
- Images/
-- 00000001_000.png
-- 00000001_001.png
-- 00000001_002.png
-- ... (approx. 15000 items)
```

The file `labels.csv` is the most important file in this dataset and has the metadata of the collected image files, stating the image index, labels, patientID, patientAge, patientGender, and the viewPosition. The sample of the dataset is given below:

| Image Index      | Finding Labels            | Follow-up # | Patient ID | Patient Age | Patient Gender | View Position | Width | Height | PixelSpacing X | PixelSpacing Y |
|------------------|---------------------------|-------------|------------|-------------|----------------|---------------|-------|--------|----------------|----------------|
| 00000001_000.png | Cardiomegaly              | 0           | 1          | 58          | M              | PA            | 2682  | 2749   | 0.143          | 0.143          |
| 00000001_001.png | Cardiomegaly\|Emphysema   | 1           | 1          | 58          | M              | PA            | 2894  | 2729   | 0.143          | 0.143          |
| 00000001_002.png | Cardiomegaly\|Effusion    | 2           | 1          | 58          | M              | PA            | 2500  | 2048   | 0.168          | 0.168          |
| 00000002_000.png | No Finding                | 0           | 2          | 81          | M              | PA            | 2500  | 2048   | 0.171          | 0.171          |

Since the `Finding Labels` contain Multiple labels like `Cardiomegaly|Emphysema` or `Cardiomegaly|Effusion`, we have a Dataset that contain multiple diseases within a single entry.

Some of the 14 commonly found Thoracic diseases and conditions are included in this dataset, which are listed below:
- Atelectasis
- Cardiomegaly
- Effusion
- Infiltration
- Mass
- Nodule
- Pneumonia
- Pneumothorax
- Consolidation
- Edema
- Emphysema
- Fibrosis
- Pleural_Thickening
- Hernia

Also the Dataset's `Finding Labels` also include the label `No Finding`, which corresponds to the image not presenting a visible symptoms of the Thoracic diseases and conditions listed above. It doesn't mean that the lung is free of Diseases and Conditions, but only the absence of the symptoms for the diseases and conditions listed

The file `train_val.txt` is another file in this dataset, which specifies which images are to be used for **training** and **validation** purposes. The file looks like this:

```bash
- 00000001_000.png
- 00000001_001.png
- 00000001_002.png
- 00000002_000.png
- 00000004_000.png
- 00000005_000.png
- ... (approx. 86500 entries of the original dataset)
```

The file `test.txt` is another file like the `train_val.txt`, which specifies which images are to be used for **testing** purposes. The filw looks like this:
```bash
- 00000003_000.png
- 00000003_001.png
- 00000003_002.png
- 00000003_003.png
- 00000003_004.png
- 00000003_005.png
- ... (approx. 25500 entries of the original dataset)
```
